Magic Square with 4 sides:

```python
problem = Problem()
problem.addVariables(range(0, 16), range(1, 16 + 1))
problem.addConstraint(AllDifferentConstraint(), range(0, 16))
problem.addConstraint(ExactSumConstraint(34), [0, 5, 10, 15])
problem.addConstraint(ExactSumConstraint(34), [3, 6, 9, 12])
for row in range(4):
    problem.addConstraint(ExactSumConstraint(34),
                        [row * 4 + i for i in range(4)])
for col in range(4):
    problem.addConstraint(ExactSumConstraint(34),
                              [col + 4 * i for i in range(4)])
solutions = problem.getSolution()
for row in range(4):
    print([solutions[row * 4 + i] for i in range(4)])
```

Magic Star:

```python
p = Problem()
domain = range(1,13)
vars = ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J", "K", "L"]

for v in vars:
    p.addVariable(v, domain)

p.addConstraint(AllDifferentConstraint(), vars)
p.addConstraint(lambda a, b, c, d, e, f, g, h, i, j, k, l: a + c + f + h == h + i + j + k == k + g + d + a 
                == e + d + c + b == b + f + i + l == l + j + g + e, vars)

solution = p.getSolution()  # You can use getSolutions() to find all solutions, but it takes a long time 
#min(solutions, key=lambda x: sum(x.values()[0] + x.values()[2] + x.values()[5] + x.values()[7])))  # for solutions  
data = [["Position"] + list(solution.keys()),
        ["Value"] + list(solution.values())]
print(tabulate(data, tablefmt='grid'))
```

The 3 smart monkeys:

```python
p = Problem(RecursiveBacktrackingSolver())
domain = range(1,4)

# Define groups in order
monkey_names = ["Zé", "Chico", "Tonho"]
actions = ["Galho", "Banana", "Pulo"]
conditions = ["Surdo", "Mudo", "Cego"]

# Add variables in order
for group in [monkey_names, actions, conditions]:
    p.addVariables(group, domain)

# Add constraints maintaining order
p.addConstraint(AllDifferentConstraint(), monkey_names)
p.addConstraint(AllDifferentConstraint(), actions)
p.addConstraint(AllDifferentConstraint(), conditions)

# Other constraints
p.addConstraint(lambda a, b: a != b, ("Zé", "Galho"))
p.addConstraint(lambda a, b: a == b, ("Surdo", "Galho"))
p.addConstraint(lambda a, b: a != b, ("Chico", "Pulo"))
p.addConstraint(lambda a, b, c: a != b != c, ("Chico", "Pulo", "Cego"))
p.addConstraint(lambda a, b, c: a != b != c, ("Tonho", "Banana", "Surdo"))

solution = p.getSolution()

# Create ordered table based on variable groups
table = []
for i in domain:
    row = []
    for group in [monkey_names, actions, conditions]:
        row.extend([k for k in group if solution[k] == i])
    table.append(row)

print(tabulate(table, headers=["Monkeys", "Actions", "Conditions"], tablefmt='grid'))
```

5 Houses:

```python
houses = Problem()
domain = range(1,6)

# Define groups in order
colors = ["red","blue","green","white","yellow"]
drinks = ["tea", "coffee", "milk", "beer", "water"]
smokes = ["prince", "bluemaster", "blends", "pall mall", "dunhill"]
animals = ["dogs","horses","birds","cats","fishs"]
nations = ["english","danish","norwegian","deutsch","swedish"]
groups = [nations, colors, drinks, smokes, animals]

# Add variables in order
for group in groups:
    houses.addVariables(group, domain)

# Add constraints maintaining order
houses.addConstraint(AllDifferentConstraint(), colors)
houses.addConstraint(AllDifferentConstraint(), drinks)
houses.addConstraint(AllDifferentConstraint(), smokes)
houses.addConstraint(AllDifferentConstraint(), animals)
houses.addConstraint(AllDifferentConstraint(), nations)

# Other constraints
houses.addConstraint(lambda a, b: a == b, ("english", "red"))
houses.addConstraint(lambda a, b: a == b, ("swedish", "dogs"))
houses.addConstraint(lambda a, b: a == b, ("danish", "tea"))
houses.addConstraint(lambda a, b: a == b - 1, ("green", "white"))
houses.addConstraint(lambda a, b: a == b, ("green", "coffee"))
houses.addConstraint(lambda a, b: a == b, ("pall mall", "birds"))
houses.addConstraint(lambda a, b: a == b, ("yellow", "dunhill"))
houses.addConstraint(lambda a: a == 3, ("milk",))
houses.addConstraint(lambda a: a == 1, ("norwegian",))
houses.addConstraint(lambda a, b: a == b + 1 or a == b - 1, ("blends", "cats"))
houses.addConstraint(lambda a, b: a == b + 1 or a == b - 1, ("horses", "dunhill"))
houses.addConstraint(lambda a, b: a == b, ("bluemaster", "beer"))
houses.addConstraint(lambda a, b: a == b, ("deutsch", "prince"))
houses.addConstraint(lambda a, b: a == b + 1 or a == b - 1, ("norwegian", "blue"))
houses.addConstraint(lambda a, b: a == b + 1 or a == b - 1, ("blends", "water"))

solution = houses.getSolution()

table = []
for group in groups:
    row = []
    for i in domain:
        row.extend([k for k in group if solution[k] == i])
    table.append(row)

print(tabulate(table, headers=domain, tablefmt='grid'))
```

Dickens:
###### From “Treinando seu C´erebro. Reader’s Digest, 2002. pg. 117.”

```python
di = Problem(BacktrackingSolver())
domain = ["D","I","C","K","E","N","S"]
rows = []

# Add variables in order
n = 0
for i in range(7):
    r = []
    for j in range(7):
        n += 1
        r.append(f"{n}")
        di.addVariable(f"{n}", domain)
    rows.append(r)

for row in rows:
    di.addConstraint(AllDifferentConstraint(), row)
    
for j in range(7):
    column = [rows[i][j] for i in range(7)]
    di.addConstraint(AllDifferentConstraint(), column)

diagonal = [rows[i][i] for i in range(7)]
anti_diagonal = [rows[i][6 - i] for i in range(7)]

di.addConstraint(AllDifferentConstraint(), diagonal)
di.addConstraint(AllDifferentConstraint(), anti_diagonal)

di.addConstraint(lambda a: a == "D", ("1",))
di.addConstraint(lambda a: a == "C", ("23",))
di.addConstraint(lambda a: a == "E", ("39",))
di.addConstraint(lambda a: a == "S", ("47",))
di.addConstraint(lambda a: a == "I", ("6",))
di.addConstraint(lambda a: a == "N", ("41",))
di.addConstraint(lambda a: a == "K", ("28",))

di.addConstraint(lambda a, b: a == b, ("3", "42"))
di.addConstraint(lambda a, b: a == b, ("9", "36"))
di.addConstraint(lambda a, b: a == b, ("5", "49"))
di.addConstraint(lambda a, b: a == b, ("22", "48"))
di.addConstraint(lambda a, b, c: a == b and b != c, ("4", "26", "43"))

di_solution = di.getSolution()
if di_solution:
    # Convert solution to 7x7 grid
    grid = []
    for i in range(7):
        row = []
        for j in range(7):
            cell_number = i * 7 + j + 1  
            row.append(di_solution[str(cell_number)])
        grid.append(row)
    print(tabulate(grid, tablefmt='grid'))
else:
    print("No solution found")
```
